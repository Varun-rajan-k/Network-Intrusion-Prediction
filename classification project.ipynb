```python
from scipy.stats import chi2_contingency

cat_cols = df.select_dtypes(include=["object","category"]).columns

for col in cat_cols:                 # testing correlation for text vs text column usinng chi-square testing

    table = pd.crosstab(df[col], df["class"])

    p = chi2_contingency(table)[1]

    print(col, "->", p)

    if p < 0.05:
        print("True")
    else:
        print("False")
```

```python
from scipy import stats

num_cols = df.select_dtypes(include=["int64","float64"]).columns

for col in num_cols:                   # testing correlation for numerical vs text column using anova test

    normal = df[df["class"]=="normal"][col]
    anomaly = df[df["class"]=="anomaly"][col]

    result = stats.f_oneway(normal, anomaly)

    print(col, "->", result.pvalue)

    if result.pvalue < 0.05:
        print("True")
    else:
        print("False")
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()          # encoding all object column to train and test because machine understand only numerical not letters

for col in df.select_dtypes(include='object').columns:
    df[col] = le.fit_transform(df[col])
```

```python
from sklearn.model_selection import train_test_split
x = df.iloc[:,0:39]
y = df.iloc[:,39]                # spliting train and test data and also traget column and other columns

x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.2,random_state = 42)
```

```python
from sklearn.preprocessing import StandardScaler
scale = StandardScaler()                #scaling for improve accuracy and treat each column fairly because if some column 
                                        #have high values the mechine gives important and gives priority to that column
```

```python
x_train_scaler = scale.fit_transform(x_train)   #  the scaling need to fit in train and transfrom and its takes all arrays
x_test_scaler = scale.transform(x_test)        # its already fits in training data so jzt need  to transform
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()      # using classification model to train and test data
```

```python
y_pred = model.predict(x_train_scaler)
```
